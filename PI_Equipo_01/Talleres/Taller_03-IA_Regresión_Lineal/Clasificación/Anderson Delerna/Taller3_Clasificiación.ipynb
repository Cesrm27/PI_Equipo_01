Ingeniería de Características y División de Datos (X, y)

```python
# Extracción de características temporales si la columna 'Date' está presente
if 'Date' in df.columns:
    df['Date'] = pd.to_datetime(df['Date'])
    df['Mes'] = df['Date'].dt.month
    df['Dia'] = df['Date'].dt.day

# Filtrar columnas numéricas que tengan variación (std > 0)
numeric_cols = df.select_dtypes(include=[np.number]).columns
var_cols = [col for col in numeric_cols if df[col].std() > 0]

# Definición de X (características) e y (variable objetivo)
target = 'Daily Mean Pb Concentration'
feature_cols = [col for col in var_cols if col != target]

X = df[feature_cols]
y = df[target]

X.head()
```

```python
y.head()
```

```python
from sklearn.model_selection import train_test_split

# División en 70% entrenamiento y 30% prueba
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=123
)
```

Modelo de Regresión Lineal

```python
from sklearn.linear_model import LinearRegression
from sklearn import metrics

# Inicialización y entrenamiento del modelo
lm = LinearRegression()
lm.fit(X_train, y_train)

print("El término de intersección del modelo lineal:", lm.intercept_)
print("Coeficientes:", lm.coef_)
```

```python
# DataFrame con los coeficientes por variable
cdf = pd.DataFrame(lm.coef_, X.columns, columns=['Coeficientes'])
cdf
```

Cálculo Manual de Errores Estándar y Estadísticos t

```python
n = X_train.shape[0]  # Cantidad de datos de entrenamiento
k = X_train.shape[1]  # Cantidad de variables
deg_f = n - k - 1     # Grados de libertad

train_pred = lm.predict(X_train)

# Cálculo del error cuadrático
train_error = np.square(train_pred - y_train)
sum_error = np.sum(train_error)

# Error estándar por cada coeficiente
se = [0] * k
for i in range(k):
    r = sum_error / deg_f
    r = r / np.sum(np.square(X_train.iloc[:, i] - X_train.iloc[:, i].mean()))
    se[i] = np.sqrt(r)

summary_df = pd.DataFrame({
    'Standard Error': se,
    'Coefficients': lm.coef_
}, index=X.columns)

summary_df['t-statistic'] = summary_df['Coefficients'] / summary_df['Standard Error']
summary_df
```

Evaluaciones Gráficas de Predicciones y Residuos

```python
# Predicciones con el conjunto de prueba
predictions = lm.predict(X_test)

print("Tipo del objeto predicho:", type(predictions))
print("Tamaño del objeto predicho:", predictions.shape)
```

```python
# Dispersión Real vs Predicho
plt.figure(figsize=(10, 7))
plt.title("Concentración Real vs. Predicha", fontsize=20)
plt.xlabel("Concentración Real de Pb", fontsize=14)
plt.ylabel("Concentración Predicha de Pb", fontsize=14)
plt.scatter(x=y_test, y=predictions, alpha=0.7)
plt.show()
```

```python
# Histograma de residuos
plt.figure(figsize=(10, 7))
plt.title("Histograma de residuos para verificar la normalidad", fontsize=20)
plt.xlabel("Residuos", fontsize=14)
plt.ylabel("Densidad", fontsize=14)
sns.histplot(y_test - predictions, kde=True)
plt.show()
```

```python
# Valores residuales vs Predichos
plt.figure(figsize=(10, 7))
plt.title("Valores residuales vs. Predichos", fontsize=20)
plt.xlabel("Predicciones", fontsize=14)
plt.ylabel("Residuos", fontsize=14)
plt.scatter(x=predictions, y=y_test - predictions, alpha=0.7)
plt.axhline(0, color='red', linestyle='--')
plt.show()
```

Modelo Basado en Árboles de Decisión e Importancia de Parámetros

```python
from sklearn import tree

# Modelo de Árbol de Decisión
tree_model = tree.DecisionTreeRegressor(max_depth=5, random_state=10)
tree_model.fit(X_train, y_train)

test_pred = tree_model.predict(X_test)

print("Mean Squared Error (MSE):", metrics.mean_squared_error(y_test, test_pred))
print("Importancia relativa de las características:", tree_model.feature_importances_)
```

```python
# Gráfico de Importancia de Características
plt.figure(figsize=(10, 6))
plt.grid(True)
plt.barh(X.columns, tree_model.feature_importances_, color='skyblue')
plt.xlabel("Importancia relativa (normalizada)", fontsize=12)
plt.ylabel("Características", fontsize=12)
plt.title("Importancia de Características - Decision Tree", fontsize=14)
plt.show()
```

Resumen Estadístico con Statsmodels (OLS)

```python
import statsmodels.api as sm

Xs = sm.add_constant(X)
stat_model = sm.OLS(y, Xs)
stat_result = stat_model.fit()

print(stat_result.summary())
```
